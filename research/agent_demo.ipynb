```python
search_tool = TavilySearchResults(
    max_results=2,
    tavily_api_key=TAVILY_API_KEY
)
```

```python
search_tool.invoke("What is the capital of France?")
```

```python
result = search_tool.invoke("Give me the latest news on AI")
result
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0,
    google_api_key=GEMINI_API_KEY
)

print("Gemini model:", llm.model)
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GEMINI_API_KEY
)
```

```python
response = llm.invoke("Explain AI agents in one sentence.")
print(response.content)
```

```python
system_prompt = """
You are a helpful research assistant.

Use the search tool when the user asks for current or factual information
that you need to look up.

Give a clear and concise final answer.
"""
```

```python
tools = [search_tool]
```

```python
agent = create_agent(
    model = llm ,
    tools = tools,
    system_prompt=system_prompt
)
```

```python
response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "Find the capital of India ?"
        }
    ]
})

print(response["messages"][-1].content)
```
